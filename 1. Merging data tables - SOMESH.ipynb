```python
import pandas as pd

# If tables are in different sheets of a single Excel file
excel_file_path = "/content/AdventureWorks Sales Data by Microsoft - BDM project.xlsx"
sheet_names = pd.ExcelFile(excel_file_path).sheet_names

dataframes = {sheet: pd.read_excel(excel_file_path, sheet_name=sheet) for sheet in sheet_names}
```

```python
dataframes.keys()
```

```python
# Start with the Sales table as the base DataFrame
combined_df = dataframes["Sales_data"]

# Merge each related table using the appropriate key
combined_df = combined_df \
    .merge(dataframes["Product_data"], on="ProductKey", how="left") \
    .merge(dataframes["Customer_data"], on="CustomerKey", how="left") \
    .merge(dataframes["Date_data"], left_on="ShipDateKey", right_on="DateKey", how="left") \
    .merge(dataframes["Reseller_data"], on="ResellerKey", how="left") \
    .merge(dataframes["Sales Order_data"], on="SalesOrderLineKey", how="left") \
    .merge(dataframes["Sales Territory_data"], on="SalesTerritoryKey", how="left")
```

```python
print(combined_df.columns)
```

```python
print(f"Original Sales row count: {len(dataframes['Sales_data'])}")
print(f"Combined DataFrame row count: {len(combined_df)}")
```

```python
# Rename columns to make it clear they belong to either Customer or Reseller
combined_df = combined_df.rename(columns={
    "City_x": "Customer City",
    "State-Province_x": "Customer State-Province",
    "Country-Region_x": "Customer Country-Region",
    "Postal Code_x": "Customer Postal Code",
    "City_y": "Reseller City",
    "State-Province_y": "Reseller State-Province",
    "Country-Region_y": "Reseller Country-Region",
    "Postal Code_y": "Reseller Postal Code"
})
```
